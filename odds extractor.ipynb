```python
"""
Odds extractor — B365, Betfair Exchange & Pinnacle uit E0 (1) t/m E0 (6)
=========================================================================
Kolommen: Date, Time, HomeTeam, AwayTeam, B365H/D/A, BFEH/BFED/BFEA, PSH/PSD/PSA
"""

import pandas as pd
import glob
import os

# ── Pas dit pad aan ───────────────────────────────────────────────────────────
BASE_DIR = r"C:\Users\semwi\FPL-Core-Insights\data\odds raw"
# ─────────────────────────────────────────────────────────────────────────────

KOLOMMEN = ['Date', 'Time', 'HomeTeam', 'AwayTeam',
            'B365H', 'B365D', 'B365A',   # Bet365
            'BFEH',  'BFED',  'BFEA',    # Betfair Exchange
            'PSH',   'PSD',   'PSA']     # Pinnacle

bestanden = sorted(glob.glob(os.path.join(BASE_DIR, "E0*.csv")))
print(f"{len(bestanden)} bestanden gevonden")

frames = []
for pad in bestanden:
    try:
        df = pd.read_csv(pad, usecols=lambda c: c in KOLOMMEN,
                         encoding='utf-8-sig')
        frames.append(df)
        print(f"  OK: {os.path.basename(pad)} — {len(df)} wedstrijden")
    except Exception as e:
        print(f"  Fout: {os.path.basename(pad)}: {e}")

alle = pd.concat(frames, ignore_index=True)
alle['Date'] = pd.to_datetime(alle['Date'], dayfirst=True, errors='coerce')
alle = alle.sort_values('Date').reset_index(drop=True)
alle = alle.dropna(subset=['B365H', 'BFEH', 'PSH'], how='all')

print(f"\nTotaal wedstrijden : {len(alle)}")
print(f"Periode            : {alle['Date'].min().date()} → {alle['Date'].max().date()}")
print(f"Betfair missend    : {alle['BFEH'].isna().sum()} wedstrijden")
print(f"Pinnacle missend   : {alle['PSH'].isna().sum()} wedstrijden\n")
print(alle.head(10).to_string(index=False))

output = os.path.join(BASE_DIR, "odds_b365_betfair_pinnacle.csv")
alle.to_csv(output, index=False)
print(f"\nOpgeslagen als: {output}")
```
