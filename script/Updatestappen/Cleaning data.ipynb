```python
import pandas as pd
import glob
import os

# ============================================================
# STAP 1: Definieer de map waar je bestanden staan
# Om een nieuw bestand toe te voegen: zorg dat het in deze map staat
# met het formaat JJJJ-JJJJ_players.csv (bijv. 2019-2020_players.csv)
# ============================================================
players_dir = r"C:\Users\semwi\FPL-Core-Insights\data\Seasonal data\players"

# ============================================================
# STAP 2: Laad alle player CSV bestanden automatisch in
# Voeg je nieuwe bestand gewoon toe aan de map — hij wordt automatisch meegenomen
# ============================================================
all_files = glob.glob(os.path.join(players_dir, "*_players.csv"))
print(f"Gevonden bestanden ({len(all_files)}):")
for f in sorted(all_files):
    print(f"  - {os.path.basename(f)}")

# ============================================================
# STAP 3: Combineer alle bestanden
# Kolommen die in nieuwe maar niet in oude bestanden zitten worden automatisch
# toegevoegd met NaN voor de seizoenen waar ze niet beschikbaar waren
# ============================================================
dfs = []
for file in sorted(all_files):
    df = pd.read_csv(file)
    season_name = os.path.basename(file).replace('_players.csv', '')
    print(f"  {season_name}: {len(df)} rijen, {len(df.columns)} kolommen")
    dfs.append(df)

combined_players = pd.concat(dfs, ignore_index=True, sort=False)

print(f"\nCombineerd totaal: {len(combined_players)} rijen | {len(combined_players.columns)} kolommen")

# ============================================================
# STAP 4: Sla op als combined bestand
# ============================================================
output_path = r"C:\Users\semwi\FPL-Core-Insights\data\player_data.csv"
combined_players.to_csv(output_path, index=False, encoding='utf-8')
print(f"\nOpgeslagen: {output_path}")
```

```python
import pandas as pd
import glob
import os
import csv

def read_csv_with_nan(file):
    """Leest CSV waarbij rijen met minder kolommen worden aangevuld met NaN."""
    with open(file, encoding='utf-8') as f:
        reader = csv.reader(f)
        rows = list(reader)
    
    header = rows[0]
    n_cols = len(header)
    
    fixed_rows = []
    for row in rows[1:]:
        if len(row) < n_cols:
            row = row + [None] * (n_cols - len(row))  # aanvullen met NaN
        elif len(row) > n_cols:
            row = row[:n_cols]  # afkappen (zou niet moeten voorkomen)
        fixed_rows.append(row)
    
    return pd.DataFrame(fixed_rows, columns=header)

# ============================================================
# STAP 1: Definieer de map
# ============================================================
matches_dir = r"C:\Users\semwi\FPL-Core-Insights\data\Seasonal data\matches"

# ============================================================
# STAP 2: Laad alle match CSV bestanden automatisch in
# ============================================================
all_files = glob.glob(os.path.join(matches_dir, "*_raw.csv"))
print(f"Gevonden bestanden ({len(all_files)}):")
for f in sorted(all_files):
    print(f"  - {os.path.basename(f)}")

# ============================================================
# STAP 3: Combineer
# ============================================================
dfs = []
for file in sorted(all_files):
    try:
        df = pd.read_csv(file)
    except Exception:
        df = read_csv_with_nan(file)
    season_name = os.path.basename(file).replace('_raw.csv', '')
    print(f"  {season_name}: {len(df)} rijen, {len(df.columns)} kolommen")
    dfs.append(df)

combined_matches = pd.concat(dfs, ignore_index=True, sort=False)
print(f"\nCombineerd totaal: {len(combined_matches)} rijen | {len(combined_matches.columns)} kolommen")

# ============================================================
# STAP 4: Sla op
# ============================================================
output_path = r"C:\Users\semwi\FPL-Core-Insights\data\match_data.csv"
combined_matches.to_csv(output_path, index=False, encoding='utf-8')
print(f"\nOpgeslagen: {output_path}")
```

```python
import pandas as pd
import os

DATA_DIR    = r"C:\Users\semwi\FPL-Core-Insights\data"
PLAYER_PATH = os.path.join(DATA_DIR, "player_data.csv")
MATCH_PATH  = os.path.join(DATA_DIR, "match_data.csv")

players = pd.read_csv(PLAYER_PATH)
matches = pd.read_csv(MATCH_PATH)

# Haal timestamp op uit match_data via match_id
# match_data heeft al leesbare timestamps na de eerdere fix
match_ts = matches[["match_id", "timestamp"]].drop_duplicates("match_id")

# Verwijder lege timestamp kolom en join vanuit match_data
players = players.drop(columns=["timestamp"], errors="ignore")
players = players.merge(match_ts, on="match_id", how="left")

# Zet timestamp voorin (na round)
cols = list(players.columns)
cols.remove("timestamp")
round_idx = cols.index("round")
cols.insert(round_idx + 1, "timestamp")
players = players[cols]

players.to_csv(PLAYER_PATH, index=False)
print(players[["round", "timestamp"]].head(5).to_string(index=False))
print(f"✅ Klaar! Ontbrekende timestamps: {players['timestamp'].isna().sum()}")
```
