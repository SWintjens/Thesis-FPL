```python
import json
import os
from datetime import datetime

MEMBERS_FILE = "members.json"


def load_members():
    if not os.path.exists(MEMBERS_FILE):
        return {}

    with open(MEMBERS_FILE, "r", encoding="utf-8") as f:
        return json.load(f)


def save_members(members):
    with open(MEMBERS_FILE, "w", encoding="utf-8") as f:
        json.dump(members, f, indent=4, ensure_ascii=False)


def add_member(name, email, starting_bankroll=100.0):
    members = load_members()

    if name in members:
        print(f"Member '{name}' bestaat al.")
        return

    members[name] = {
        "email": email,
        "starting_bankroll": float(starting_bankroll),
        "current_bankroll": float(starting_bankroll),
        "active": True,
        "created_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "sheet_id": None,
        "sheet_url": None
    }

    save_members(members)
    print(f"Member '{name}' toegevoegd.")


def remove_member(name):
    members = load_members()

    if name not in members:
        print(f"Member '{name}' bestaat niet.")
        return

    del members[name]
    save_members(members)
    print(f"Member '{name}' verwijderd.")


def deactivate_member(name):
    members = load_members()

    if name not in members:
        print(f"Member '{name}' bestaat niet.")
        return

    members[name]["active"] = False
    save_members(members)
    print(f"Member '{name}' gedeactiveerd.")


def activate_member(name):
    members = load_members()

    if name not in members:
        print(f"Member '{name}' bestaat niet.")
        return

    members[name]["active"] = True
    save_members(members)
    print(f"Member '{name}' geactiveerd.")


def update_bankroll(name, new_bankroll):
    members = load_members()

    if name not in members:
        print(f"Member '{name}' bestaat niet.")
        return

    members[name]["current_bankroll"] = float(new_bankroll)
    save_members(members)
    print(f"Bankroll van '{name}' geüpdatet naar {new_bankroll}.")


def set_member_sheet(name, sheet_id=None, sheet_url=None):
    members = load_members()

    if name not in members:
        print(f"Member '{name}' bestaat niet.")
        return

    members[name]["sheet_id"] = sheet_id
    members[name]["sheet_url"] = sheet_url
    save_members(members)
    print(f"Sheet info opgeslagen voor '{name}'.")


def list_members(active_only=False):
    members = load_members()

    if not members:
        print("Geen members gevonden.")
        return

    for name, info in members.items():
        if active_only and not info.get("active", False):
            continue

        print(
            f"- {name} | "
            f"email: {info['email']} | "
            f"bankroll: {info['current_bankroll']} | "
            f"active: {info['active']}"
        )


def get_member(name):
    members = load_members()
    return members.get(name)


# Voorbeeldgebruik
add_member("Swin", "sem.wintjens@home.nl", 100)
add_member("Jan", "jan@email.com", 150)
add_member("Patrick", "patrick@email.com", 1000)

list_members()
```
